```python
import numpy as np
import PIL.Image as Image
import matplotlib.pyplot as plt
%config InlineBackend.print_figure_kwargs = {'bbox_inches': None}
from IPython.display import IFrame
from IPython.display import HTML
from IPython.display import display
from matplotlib.animation import FuncAnimation

# IPython.display アニメーションの定義
import matplotlib
matplotlib.rcParams['animation.embed_limit'] = 70 # アニメの容量制限を70MBにアップ

def makeAnimation(frames,step=1):
    fig = plt.figure(figsize=(frames[0].shape[1]/100, frames[0].shape[0]/100), dpi=100,frameon=False)
    ax = fig.add_axes([0.,0.,1.,1.])
    plt.axis('off')
    patch = plt.imshow(frames[0][:,:,::-1])

    def animate(i):
        patch.set_data(frames[i][:,:,::-1])

    anim = FuncAnimation(plt.gcf(), animate, frames=len(frames), interval=1000/(120.0/step))
    display(HTML(anim.to_jshtml()))
```

```python
base = cv2.imread("ImageA.png",1)
# base = cv2.imread("myobject.png",1)  # 自分の用意した画像名に書き換える
base_gray = cv2.cvtColor(base, cv2.COLOR_BGR2GRAY)
(h,w) = base_gray.shape
flag = (base_gray[h-1,w-1] < base_gray[0,0]) # 背景の濃度
base0 = base[0,0] if flag else base[h-1,w-1]
gray0 = base_gray[0,0] if flag else base_gray[h-1,w-1]
display(Image.fromarray(base[:,:,::-1]))
print("図１　原画像")
ret,cimg = cv2.threshold(base_gray,gray0+5,255,cv2.THRESH_BINARY)
# 穴埋めして少し大きくする
kernel = np.ones((5,5),np.uint8)
cimg = cv2.morphologyEx(cimg, cv2.MORPH_OPEN, kernel)
# cimg = cv2.dilate(cimg,kernel,iterations = 1)
# 白領域をラベリング
nLabels, labelImages, data, center = cv2.connectedComponentsWithStats(cimg)
bindex = np.argmax(data[:,4]) # 最大の城領域の番号
limg = ((labelImages!=bindex)*255).astype(np.uint8)
# 太めた画像の輪郭追跡
if list(cv2.__version__)[0] == '4':
    contours,hierarchy = cv2.findContours(limg, 1, 2) # OpenCV 4.0 の場合
else:
    imgEdge,contours,hierarchy = cv2.findContours(limg, 1, 2) # OpenCV 3.x の場合
# もっとも長い輪郭線を選んで描画しマスク画像を作成
maxcon = np.argmax( [len(con) for con in contours])
con = contours[maxcon]
x,y,ow,oh = cv2.boundingRect(con) # 対象を囲む矩形を決定
mask = np.zeros_like(base_gray)
mask = cv2.drawContours(mask, [con], 0, 1, -1) #マスク画像
# mask を使って、オブジェクト以外を黒で描画
object = base*(np.where(mask==1,1,0)[:,:,np.newaxis].astype('uint8'))
# オブジェクトを矩形で切り出し
object = object[y:y+oh,x:x+ow]
display(Image.fromarray(object[:,:,::-1]))
print("図２　オブジェクト")

if ow > 135 or oh > 135:
    print(f"縦{oh}, 横{ow}")
    print("警告: オブジェクトのサイズが大きすぎます。MagicaVoxelに戻って、オブジェクトのサイズを少し小さくしてください。")
```

# Shi-Tomasi コーナー検出
この実験で用いるLucus-Kanade 法では，コーナーなどの追跡しやすい特徴点のみを追跡する． よい特徴として，Harris コーナーや，Shi-Tomasiのコーナーがよくつかわれる． OpenCV の goodFeaturesToTrack()関数は，Shi-Tomasiのコーナー検出が実装されている．

オプティカルフロー検出を試みる前に，goodFeaturesToTrack()でどのような箇所が検出されるか見ておこう．

```python
GFT_PARAMS = dict( maxCorners = 30,
                       qualityLevel = 0.01,
                       minDistance = 9,
                       blockSize = 3 )

corners = cv2.goodFeaturesToTrack(base_gray,**GFT_PARAMS).astype(np.intp)

num_features = len(corners)
print(f"検出された良い特徴点の数: {num_features}")
if num_features < 10:
    print("警告: 検出された特徴点が10個未満です。画像の選択をやり直すか、MagicaVoxelでオブジェクトを変更してください。")

gray = cv2.cvtColor(base_gray,cv2.COLOR_GRAY2BGR)
for i in corners:
    x,y = i.ravel()
    cv2.circle(gray,(x,y),3,(0,0,255),-1)

display(Image.fromarray(gray[:,:,::-1]))
print("図３　Good Features = Shi-Tomasiコーナー")
```

<font color='red'>もし，良い特徴の数が少ないようであれば，入力図形を変更すること．</font> 10個以上あればOK

# 疑似動画生成１　並進運動
レンダリングに３０秒程度かかります

```python
%%time
movie = []

bimg = np.zeros((320,640,3),np.uint8)

for line in range(bimg.shape[0]): # 背景を作る
    bimg[line,:] = base[line,0]

mask = np.where(object>0,0,255).astype(np.uint8)
for i in range(5,635-ow): #
    img = bimg.copy()
    sh = 160-int(oh/2)
    crip = img[sh:sh+oh,i:i+ow,:]
    crip = cv2.bitwise_and(crip,mask,crip)
    img[sh:sh+oh,i:i+ow,:] = cv2.add(crip,object)
    img = cv2.GaussianBlur(img, (3, 3), 0)
    movie.append(img)

makeAnimation(frames=movie)
```

# 疑似動画生成２　回転運動

```python
%%time
(oh,ow)=object.shape[:2]
rimg = np.zeros((480,480,3),np.uint8)
offset = -30
ocx,ocy = ow-offset,offset # オブジェクトを囲むボックスの右上から内側に(offset,offset)入った点を中心にする
cx,cy = 240,240 # 図形全体の中心
dx,dy = cx - ocx, cy - ocy #
rimg[dy:dy+oh,dx:dx+ow] = object # 回転の初期位置

base480 = cv2.resize(base,(480,480))
bimg = np.zeros_like(base480)
for line in range(len(base480)): # 背景を作る
    bimg[line,:] = base480[line,0]

amovie = []
for angle in range(360): #
    trans = cv2.getRotationMatrix2D((cx,cy), angle, scale=1.0)
    oimg = cv2.warpAffine(rimg, trans, (480,480))
    mask = np.where(oimg>0,0,255).astype(np.uint8)
    bimg1 = bimg.copy()
    bimg1 = cv2.bitwise_and(bimg1,mask,bimg1)
    img = cv2.add(oimg,bimg1)
    img = cv2.GaussianBlur(img, (3, 3), 0)
    amovie.append(img)
# 360度を超える角度の場合，（angle -360度）と同じになる．少し余分に作っておく
for angle in range(360,405):
    amovie.append(amovie[angle-360])

makeAnimation(frames=amovie)
```

# Lucas-Kanade法によるオプティカルフロー追跡
## 並進移動の場合

```python
# Parameters for lucas kanade optical flow
LK_PARAMS = dict( winSize  = (10,10),
                  maxLevel = 5,
                  criteria = (cv2.TERM_CRITERIA_MAX_ITER|cv2.TERM_CRITERIA_EPS | cv2.TERM_CRITERIA_COUNT, 20,.05))

# Create some random colors
color = np.random.randint(0,255,(100,3))
# Take first frame and find corners in it

def showOF(anim, start=100,step=1):
    old_frame = anim[start]
    old_gray = cv2.cvtColor(old_frame, cv2.COLOR_BGR2GRAY)

    # 初期追跡特徴点の抽出
    p0 = cv2.goodFeaturesToTrack(old_gray, mask=None, **GFT_PARAMS)
    mask = np.zeros_like(old_frame)
    frames = [old_frame]

    # 特徴点数の推移を記録するリスト
    points_count_history = [(100,len(p0))]
    print(f"初期（{start}フレーム目）の追跡対象特徴点は {len(p0)}個")

    for i in range(start+step, len(anim), step):
        frame = anim[i].copy() # 元の画像を汚さないようcopyを推奨
        frame_gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)

        # 光学的流動の計算
        p1, st, err = cv2.calcOpticalFlowPyrLK(old_gray, frame_gray, p0, None, **LK_PARAMS)

        if p1 is None or st is None:
            print(f"Frame {i}: 完全に追跡に失敗しました")
            break

        # 有効な点のみ抽出 (st==1 が追跡成功)
        good_new = p1[st == 1]
        good_old = p0[st == 1]

        # 特徴点数を記録
        current_count = len(good_new)
        if current_count != points_count_history[-1][1]:
              points_count_history.append((i,current_count))
              print(f"{i}フレーム目，特徴点 {points_count_history[-2][1]-current_count}個ロスト，継続追跡{current_count}個")

        # 描画処理
        for j, (new, old) in enumerate(zip(good_new, good_old)):
            a, b = new.ravel().astype(int)
            c, d = old.ravel().astype(int)
            mask = cv2.line(mask, (a, b), (c, d), color[j % 100].tolist(), 2)
            frame = cv2.circle(frame, (a, b), 5, color[j % 100].tolist(), -1)

        img = cv2.add(frame, mask)
        frames.append(img)

        # 更新：次のループへの準備
        old_gray = frame_gray.copy()
        p0 = good_new.reshape(-1, 1, 2)

        # 特徴点が少なくなった場合の再抽出
        if len(p0) < 6:
             print(f"Frame {i}: 特徴点が5個以下（{len(p0)}個）に減少したため、再抽出しました。")
             p0 = cv2.goodFeaturesToTrack(old_gray, mask=None, **GFT_PARAMS)
             # 再抽出直後は軌跡(mask)をリセットしたい場合はここで mask = np.zeros_like(old_frame)

    # 推移の表示
    print("-" * 30)
    print("特徴点数の推移:", points_count_history)
    print("-" * 30)

    makeAnimation(frames=frames, step=step)
```

#### 上でおおまかに限界をつかむことができたら、STEPを細かく刻んで限界を調べよう。
次のセルで、2フレーム間のフローを描画する関数を定義している。

```python
# ２つのフレームのマッチング
def drawCorr(anim=movie,A=0,B=1):

    display(Image.fromarray(anim[A][:,:,::-1]))
    print("フレームA")
    display(Image.fromarray(anim[B][:,:,::-1]))
    print("フレームB")

    old_frame = anim[A].copy()
    old_gray = cv2.cvtColor(old_frame, cv2.COLOR_BGR2GRAY)
    new_frame = anim[B].copy()
    new_gray = cv2.cvtColor(new_frame, cv2.COLOR_BGR2GRAY)

    # drawing colors
    yellow = (255,255,0)
    red = (255,0,0)
    blue = (0,0,255)

    # Take first frame and find corners in it
    p0 = cv2.goodFeaturesToTrack(old_gray, mask = None, **GFT_PARAMS)

    # Create a mask image for drawing purposes
    mask = np.zeros_like(old_frame)

    # calculate optical flow
    p1, st, err = cv2.calcOpticalFlowPyrLK(old_gray, new_gray, p0, None, **LK_PARAMS)

    # Select good points
    good_new = p1[st==1]
    good_old = p0[st==1]

    old_frame[...,0]=0
    new_frame[...,1]=0

    blended = cv2.addWeighted(src1=old_frame,alpha=1,src2=new_frame,beta=1,gamma=0)

    # draw the tracks
    for i,(new,old) in enumerate(zip(good_new,good_old)):
        a,b = new.ravel().astype(int)
        c,d = old.ravel().astype(int)
        mask = cv2.line(mask, (a,b),(c,d), yellow, 1)
        frame = cv2.circle(blended,(a,b),2,red,-1)
        frame = cv2.circle(frame,(c,d),2,blue,-1)
    img = cv2.add(frame,mask)

    display(Image.fromarray(img))
    print("特徴点マッチング結果")
```

## フロー追跡の限界調査
先におおまかにつかんだTESTSTEPをベースにして、次のプログラムのSTEPを少しずつ変えて限界を調べよう。

```python
# フレーム間隔
START = 100
STEP  = 99 #@param {type:"slider", min:1, max:256, step:1}
drawCorr(anim=movie,A=START,B=START+STEP)  # STEP目のフレームと 2*STEP目のフレームとの間のオプティカルフロー
```

# 回転運動の場合

並進運動と同様に限界を調査しよう。

ただし，Lucas-Kanade法は並進運動を前提としたアルゴリズムであり，回転運動は想定外である。また、空と地面の境界を越える時に特徴点近傍の様相は大きく変化するので多くの特徴点の追跡はそこで途絶えてしまう。そのためすべての点の追跡は望めない。

そこで，次の事項を調査し、報告せよ。

1. 初期特徴点の半数が追跡できなくなる角度（STEPが角度そのもの）
2. 上のSTEPに設定したときに、最初に特徴点を見失ってしまうフレーム数


```python
# 角度間隔
ANGLESTEP  = 15 #@param {type:"slider", min:1, max:60, step:1}
showOF(anim=amovie,start=0,step=ANGLESTEP)
```

# 補足
- このプログラムでは対応する特徴点を直線で結んでいるだけなので、回転運動だからといって、軌跡が円弧状に描かれることはなく、必ず折れ線になります。例えば、ステップが45度なら理想的な軌跡はが８角形になります。  
- この実験のプログラムのSTEP数は、並進運動と回転運動とでは、意味も単位も異なります。よって、その値によって並進と回転を比較するのはナンセンスです。
- さらに言うと、オプティカルフロー追跡の理論は通常は並進運動を仮定して組み立てられていますので、回転運動には弱くて当然です。並進を前提とした追跡法でも変化が小さければ回転にもある程度までは対応できます。その限界を確かめるのがこの実験（回転の実験）の狙いです。

# おまけ
以下のプログラムは，密なオプティカルフロー追跡

色の違いは移動方向の違いを表す

```python
cap = movie
(h,w) = cap[0].shape[:2]

NOMOVE = 0.9

def draw_flow(img, flow, step=16):
    global w,h
    y, x = np.mgrid[step/2:h:step, step/2:w:step].reshape(2,-1).astype(int)
    fx, fy = flow[y,x].T
    lines = np.vstack([x, y, x+fx, y+fy]).T.reshape(-1, 2, 2)
    lines = np.int32(lines + 0.5)
    vis = cv2.cvtColor(img, cv2.COLOR_GRAY2BGR)
    mag, ang = cv2.cartToPolar(flow[...,0], flow[...,1])
    hsv = np.zeros((h, w, 3), np.uint8)
    hsv[...,0] = ang*(180/np.pi/2)
    hsv[...,1] = 255
    # hsv[...,2] = cv2.normalize(mag,None,0,255,cv2.NORM_MINMAX)
    hsv[...,2][mag>NOMOVE] = 255
    bgr = cv2.cvtColor(hsv, cv2.COLOR_HSV2BGR)
    for (x1, y1), (x2, y2) in lines:
        color = (int(bgr[y1,x1,0]),int(bgr[y1,x1,1]),int(bgr[y1,x1,2]))
        # cv2.circle(vis, (x1, y1), 1, color, -1)
        cv2.line(vis,(x1,y1),(x2,y2), color ,1)
    return vis

prev = cap[0]
prevgray = cv2.cvtColor(prev, cv2.COLOR_BGR2GRAY)

offrames = []
for img in movie[1:]:
    gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
    flow = cv2.calcOpticalFlowFarneback(prevgray, gray, None, pyr_scale = 0.5,levels = 3,
             winsize = 15, iterations = 3, poly_n = 5, poly_sigma = 1.1, flags = 0)
    prevgray = gray
    mag, ang = cv2.cartToPolar(flow[...,0], flow[...,1])
    dimg = draw_flow(gray, flow, step=8)
    offrames.append(dimg)

# 動画表示
makeAnimation(frames=offrames)  #
```
